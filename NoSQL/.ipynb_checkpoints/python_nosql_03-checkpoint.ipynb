```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017")

db = client.sample_mfilx
movies = db.movies
```

```python
# for group in movies.aggregate([
#     {"$group":{"_id": "$directors", "count":{"$sum": 1}}}
# ]):
#     print(group)

group_pipeline = {"$group": {"_id": "$directors", "count": {"$sum": 1}}}
sort_pipeline = {"$sort": {"count": -1}}

pipelines = [group_pipeline, sort_pipeline]

for group in movies.aggregate(pipelines) :
    print(group)
```

```python
#match
# pipleline = [{"$match": {"genres": "Action"}}, {"$limit":1)]

# for movie in movies.aggregate(pipeline) :
#     print(movie)

pipeline = [{"$match": {"genres": "Action"}}, {"$limit": 1}]

for movie in movies.aggregate(pipeline) :
    print(movie)
```

```python
# for movie in movies.aggregate([
#     {"#sort":{"title":1}}, {"$limit":3}
# ])

for movie in movies.aggregate([
    {"$sort": {"title": 1}},
    {"$limit": 3}
]) :
    print(movie)
```

```python
# $match,$group,$sort,$limit, $project
#왜 어떤 필드명 앞에은 $가 붙고, 또 붙지 않을까?
#$는 해당 필드명 내 컬럼 안에 있는 값들을 개별적으로 조회해서 어떠한 연산처리를 해야하는 경우 붙는다.
#이미 그춥화 되어있는 컴럼 그 자체를 통으로 가져와서 연산처리를 해야하는 경우 붙지 않음.

for movie in movies.aggregate([
    {"$project": {"_id" : 0,"title" :1, "genres" :1}}, {"$limit":1}
]): 
    print(movie)
```

```python
for movie in movies.aggregate ([
    {"$unwind": "$genres"},
    {"$limit":3}
]) :
    print(movie)
```

```python
# for group in movies.aggregate([
#     {"$group": {"_id": "$directors","average_rating":{"$avg":"$imdb.rating"}}}, 
#     {"$limit: 5}
# ]):
#     print(group)

for group in movies.aggregate([
    {"$group": {"_id": "$directors", "average_rating": {"$avg": "$imdb.rating"}}},
    {"$limit": 5}
]) :
    print(group)
```

```python
for item in movies.aggregate([
    {"$group": {"_id": None,"avg_length": {"$avg":"$runtime"}}}
]) :
    print(item["avg_length"])
```

```python
for genre in movies.aggregate([
    {"$unwind": "$genres"}, 
    {"$group" : {"_id": "$genres" , "count": {"$sum":1}}}
]) :
    print("장르:",genre["_id"], "영화 수 :",genre["count"])
```

```python
for movie in movies.aggregate([
    {"$unwind": "$countries", "count": {"$sum":1},
]) :
    
    print(movie)
```
