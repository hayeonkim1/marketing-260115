RDBMS : 관계형 데이터베이스 관리 시스템
    - 현존하는 가장 오래된 데이터베이스 관리 시스템
    - 안정적, 가성비, 대중적
    - 시대 변화 -> 문제 지적발생
    - MySQL활용 RDBMS -> 가져와야하는 데이터의 형태가 다 다름
    정해진 패턴을 기반으로 데이터 수집, 저장// 데이터 다양 & 방대해짐 => 빅데이터 시대
    -> 패턴(스키마)를 기반으로 테이블에 데이터를 차곡차곡 저장하는 시대가 아님

NoSQL : Not only SQL (전통적인 SQL문법만 의존하는 관계형 데이터베이스 관리 시스템 탈피)
    - 고정된 Scheme 포기. / JOIN 문법자체가 존재불가능 

    - RDBMS: 최초에 한번 혹은 아주 가끔씩만 데이터 저장 / 이미 저장된 데이터를 수정없이 그냥 읽어오는 업무프로세스에 많이 사용 => MySQL, Oracle, PostgreSQL, SLQlite

    - NoSQL: 빈번하게 데이터의 입출력이 가능해야하는 상황(db -> ecommerce /멤버십 관리/ 금융권)
        => MongoDB
        -> JSON 형태를 띄고 있음 => JAVASCRIPT Object Notation
                - jacascript: 웹 브라우저가 인식할 수 있는 유일한 프로그래밍 언어
                - BSON문법체계 => 빅데이터

RDBMS : DB > TABLE > SCHEME > ROW > COLUMN > FIELD > PK | FK
NoSQL: DB > COLLECTION > DOCUMENT(KEY,VALUE) > FIELD > _id(=PK)
