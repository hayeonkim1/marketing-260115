```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017")

#값 생성
db = client["test"]
users = db.users
#컬렉션에 값넣기
data = {"name": "David", "age": "20"}
result = users.insert_one(data)

print(result) 
```

```python
client.close() #메모리 누수를 방지하기 위한 목적
```

```python
#close()룰 하지않아도 끝내게하기 with 절
with MongoClient("mongodb://localhost:27017") as client:
    db = client["test"]
    users = db.users
    data = {"name": "David", "age": "20"}
    result = users.insert_one(data)
```

```python
#mongodb에 데이터 삽입
with MongoClient("mongodb://localhost:27017") as client:
    db = client["test"]
    users = db.users
    data = [
    {"name":"Kim", "age":25},
    {"name":"Lee", "age":35},
    {"name":"Park", "age":40},
    ]
    result = users.insert_many(data)
```

```python
#값 조회하기

client = MongoClient("mongodb://localhost:27017")

db = client["test"]
collection = db.users

# document = collection.find_one({"name":"David"})

documents = collection.find()   #=> 전체를 찾아올때 이터러블한 속성의 값이므로 반복문을 통해 하나하나씩 찾아와야 컬렉션의 복수의 값들을 찾아올 수 있음.

for document in documents :
    print(document)
```

```python
#값 수정하기
client = MongoClient("mongodb://localhost:27017")

db = client["test"]
collection = db.users

result = collection.update_one({"name":"David"},{"$set":{"age":30}}) #{"name":"David"}이면 "$set":{"age":30} 이렇게 업데이트($set) 해라
print(result)
```

```python
#값 수정하기(2)
client = MongoClient("mongodb://localhost:27017")

db = client["test"]
collection = db.users
filter = {"name":"David"}
update = {"$set":{"age":30}}

result = collection.update_one(filter,update) 
print(result, result.matched_count, result.modified_count)
```

```python
client = MongoClient("mongodb://localhost:27017")

db = client["test"]
collection = db.users

filter = {"age": {"$gt":30}}
update = {"$set":{"is_available":True}} 
result = collection.update_many(filter,update)

print("matched Count:", result.matched_count, "modified Count:", result.modified_count)
```

```python
## 단일 값 제거하기

client = MongoClient("mongodb://localhost:27017")

db = client["test"]
collection = db.users

filter = {"name":"Park"}
result = collection.delete_one(filter)

print("Deleted Count:", result.deleted_count)
```

```python
## 복수 값 제거하기

client = MongoClient("mongodb://localhost:27017")

db = client["test"]
collection = db.users

filter = {"age": {"$lte":30}}
result = collection.delete_many(filter)
print("Deleted Count:", result.deleted_count)
```
