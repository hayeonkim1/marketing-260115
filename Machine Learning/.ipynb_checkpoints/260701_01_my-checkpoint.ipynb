```python
title_mapping = {"Mr": 0, "Miss": 1, "Mrs": 2, "Master": 3, "Lady": 4, "Don": 5, "Rev": 6, "Dr": 7, "Major": 8, "Sir": 9, "Col": 10, "Capt": 11, "Jonkheer": 12}
df["Initial"] = df["Initial"].map(title_mapping)
```

```python
df.head()
```

```python
import pandas as pd

train = pd.read_csv("titanic/train.csv")
test = pd.read_csv("titanic/test.csv")

ntrain = train.shape[0]
ntest = test.shape[0]
```

```python
ntrain #학습데이터 행 수
```

```python
ntest #훈련데이터 행 수
```

```python
#High Chance Feature : 생존확률이 높은 개별적인 피처들을 모아서 상대적으로 생존확률을 높인 집단의 피처: 보다 확룰이 높은 요소들의 Feature들을 하나로 모아놓은 새로운 Features
#low Chance Feature: 보다 확룰이 낮은 요소들의 Feature들을 하나로 모아놓은 새로운 Features
```

```python
# 앞에서 feature engineering 에서 발견한 항목들을 먼저 발견하고
#생존확률이 높은 조건들만 가지고 따로 df만들어보기
#생존확률이 높은 조건들 순으로 모아서 번호 매기기

df["HighChance"] = 0
df.loc[(df["Age"] == 0), "HighChance"] = 1
df.loc[(df["Pclass"] == 1) & (df["Initial"] == 2), "HighChance"] = 2
df.loc[(df["Pclass"] == 1) & (df["Initial"] == 1), "HighChance"] = 3
df.loc[(df["Pclass"] < 3) & (df["Age"] == 2) & (df["Sex"] == 1), "HighChance"] = 4
df.loc[(df["Initial"] == 4), "HighChance"] = 5
df.loc[(df["Pclass"] < 3) & (df["Age"] < 2) & (df["Sex"] == 1), "HighChance"] = 6
```

```python
train = df[:ntrain]
```

```python
train.head()
```

```python
import plotly.express as px

train_high = train.copy()

train_high["Outcome"] = train_high["Survived"].map({0: "Dead", 1: "Survived"})

fig = px.histogram(
    train_high,
    x="HighChance",
    color="Outcome",
    barmode="overlay",
    title="HighChance Distribution By Survival Status (Overlay)"
)

fig.show()
```

```python
df["LowChance"] = 0
df.loc[(df["Sex"] == 0) & (df["Family"] == 1), "LowChance"] = 1
df.loc[(df["Sex"] == 0) & (df["Family"] == 1) & (df["Pclass"] == 3), "LowChance"] = 2
```

```python
df.shape
```

```python
train = df[:ntrain]
```

```python
train.head()
```

```python
import plotly.express as px

train_low = train.copy()

#매핑하기
train_low["Outcome"] = train_low["Survived"].map({0: "Dead", 1: "Survived"})


fig = px.histogram(
    train_low,
    x="LowChance",
    color="Outcome",
    barmode="overlay",
    title="LowChance Distribution By Survival Status(Overlay)"
)

fig.show()
```
