```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support import expected_conditions as EC #-> 알림창 같은 unexpected condition 방지를 위해
from selenium.webdriver.support.ui import WebDriverWait
from webdriver_manager.chrome import ChromeDriverManager
import time

# 상수값으로서 자주쓰는 변수 목록을 보기쉽게 위쪽에 정리
URL = "https://moolix.netlify.app/"

service = Service(ChromeDriverManager().install())
options= Options()

# options.add_argument("--headless")
# options.add_argument("--disable-gpu")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--window-size=1920,1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")
options.add_argument("--no-snadbox")

driver= webdriver.Chrome(service=service, options=options)

driver.get(URL)
time.sleep(3)



#로그인/로그아웃 자동화
target_btn = driver.find_element(By.ID, "btnLogin")

target_btn.click()  #--> 찾은 요소를 클릭해라
time.sleep(2)

#placeholder 값 제거하고 새로운 값 넣기
id_element = driver.find_element(By.ID, "authEmail")
id_element.clear()
id_element.send_keys("hayeonkim1@gmail.com")

pw_element = driver.find_element(By.ID, "authPw")
pw_element.clear()
pw_element.send_keys("1234")

login_btn = driver.find_element(By.ID,("authSubmit"))
login_btn.click()

try :
    WebDriverWait(driver, 5).until(EC.alert_is_present()) #-> 5초간 기다릴동안 알림창이 나오면 실행
    alert = driver.switch_to.alert
    alert.accept()

except : #-> 5초 기다리는 동안 알림창이 안나오면 실행
    print("Alert None")


print(target_btn.text)


driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
from urllib.request import urlretrieve
import time

# 상수값으로서 자주쓰는 변수 목록을 보기쉽게 위쪽에 정리
URL = "https://davelee-fun.github.io/"

service = Service(ChromeDriverManager().install())
options= Options()

# options.add_argument("--headless")
# options.add_argument("--disable-gpu")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--window-size=1920,1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")
options.add_argument("--no-snadbox")

driver= webdriver.Chrome(service=service, options=options)
driver.get(URL)


sources = []
#source = list()

elements = driver.find_elements(By.CSS_SELECTOR, "div.maxthumb img")
for element in elements :
    sources.append(element.get_attribute("src"))

for index, source in enumerate(sources) :
    urlretrieve(source, "image/" + str(index) + "." + source.split(".")[-1])

driver.quit()
```
