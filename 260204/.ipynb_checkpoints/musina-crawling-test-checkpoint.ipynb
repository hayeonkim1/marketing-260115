```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from webdriver_manager.chrome import ChromeDriverManager
import pandas as pd
import time
import re 

URL = "https://www.musinsa.com/main/musinsa/"

service = Service(ChromeDriverManager().install())
options= Options()

# options.add_argument("--headless")
# options.add_argument("--disable-gpu")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--window-size=1920,1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")
options.add_argument("--no-snadbox")

driver= webdriver.Chrome(service=service, options=options)
driver.get(URL)
time.sleep(3)

#상단 검색바 클릭
search_button = driver.find_element(By.CLASS_NAME, "Searchbar__TextFieldWrap-sc-dzagxs-2")
search_button.click()
time.sleep(5)



#인기 검색어 1위 키워드 서칭 -니트
popular_keyword = driver.find_element(By.CLASS_NAME, "search-home-popular-wrap")
items = popular_keyword.find_elements(By.TAG_NAME, "li")
target_keyword = items[0].text
time.sleep(3)


#니트 키워드 입력> 니트 상품 PLP
search_input = driver.find_element(By.CLASS_NAME, "search-home-search-bar-keyword")
search_input.clear()
search_input.send_keys(target_keyword)
search_input.send_keys(Keys.RETURN)
time.sleep(2)


#니트 관련 상품 목룍 > 인기 상품 5개만 선정
#스크롤

product_data = list()
for _ in range(2) :
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(3)
    
    products = driver.find_elements(By.CSS_SELECTOR,"div.sc-cpclqO.jrLXZA")[:5]
    
    for product in products :
        try:
            product_info = product.find_element(By.CSS_SELECTOR, "div.sc-itBLYH.cDYUnS")
            brand = product_info.find_element(By.CSS_SELECTOR,"span.text-etc_11px_semibold" ).text.strip()
            name = product_info.find_element(By.CSS_SELECTOR, "span.text-body_13px_reg").text.strip()
            url = product_info.find_element(By.CSS_SELECTOR, "a.gtm-select-item").get_attribute("href")
            review_infos = product.find_elements(By.CSS_SELECTOR, "span.text-etc_11px_reg.text-yellow.font-pretendard")
            review_rating = float(review_infos[0].text.strip()) if len(review_infos) > 0 else ""
            review_count_before = review_infos[1].text.strip() if len(review_infos) > 1 else ""
            #정규표현식으로 숫자가 아닌 값 다 날리기 -> 정수가 아닌 값을 찾아서 없애라
            review_count = int(re.sub( r"[^\d]","", review_count_before)) #-> 정수로 형변환
            product_data.append({"브랜드명" : brand, "상품명": name, "URL" : url,"평점" :review_rating,"리뷰수": review_count})

        except Exception as e:
            print("데이터 크롤링 실패:", e)
            continue

df = pd.DataFrame(product_data)
df_top10 = df.sort_values(by="리뷰수", ascending=False).head(10).reset_index(drop=True)  #리뷰수라는 column으로 내림차순 정렬 but, 리뷰수를 인덱스 값으로 쓰고싶지는 않음  
driver.quit()

df_top10
```

```python
#인기상품 5개 각각의 상세페이지(PDP) 접속
#각 상품별 리뷰(댓글) 목록 확인
#해당 리뷰 수집
#엑셀 저장
```

```python
from openpyxl import Workbook

driver= webdriver.Chrome(service=service, options=options)
all_reviews = list()

for idx, row in df_top10.iterrows() : #-> 한 행씩 값을 읽기를 반복하겠다는 매서드 함수
    print(f"{idx+ 1}번째 상품리뷰 수집 중 :", row["상품명"])  #-> 중간에 오류지점을 추적하기 위함
    driver.get(row["URL"])
    time.sleep(2)

    try : 
        review_tab = driver.find_element(By.CSS_SELECTOR,"button[data-button-id='prd_review_tab']")
        review_tab.click()
        time.sleep(2)

        review_elements = driver.find_elements(By.CSS_SELECTOR, "div.ExpandableContent__Container-sc-1951ro-0 > div.ExpandableContent__ContentContainer-sc-1951ro-1 span.text-body_13px_reg.w-full.text-black.font-pretendard")
        for review in review_elements :

            text = (review.text or "").strip()

            if not text : 
                continue
            
            all_reviews.append({
                "브랜드" : row["브랜드명"],
                "상품명" : row["상품명"],
                "URL": row["URL"],
                "리뷰수": row["리뷰수"],
                "리뷰" : review.text.strip(),
                "평점" : row["평점"]
            })
      
    except Exception as e :
        print("리뷰수집 실패 :", e)
        continue


driver.quit()

wb = Workbook()
ws = wb.active
ws.title = "무신사 리뷰"
ws.append(["브랜드","상품명","URL","리뷰수","리뷰","평점"])  #-> 행에 목록 추가: []

for item in all_reviews : 
    ws.append([item["브랜드"],item["상품명"],item["URL"],item["리뷰수"],item["리뷰"],item["평점"]])

file_name = "musinsa_top10_reviews.xlsx"
wb.save(file_name)

print("엑셀파일 저장완료:", file_name)
```

```python
import pandas as pd
import re
import hashlib
from sqlalchemy import create_engine, text

EXCEL_PATH = "musinsa_top10_reviews.xlsx"
HOST = "127.0.0.1"
PASSWORD = "Hayeon123!"
USER = "root"
PORT = 3306
DB = "musinsa_db" 

df = pd.read_excel(EXCEL_PATH)

ENGINE = create_engine(f"mysql+pymysql://{USER}:{PASSWORD}@{HOST}:{PORT}/{DB}?charset=utf8mb4", pool_pre_ping= True, pool_recycle=3600)

df.columns = [str(c).strip() for c in df.columns] #리스트안에 담겨있음 [브랜드, 상품명, url, 리뷰수, 리뷰, 평점]
needed = ["브랜드","상품명", "URL", "리뷰수", "리뷰", "평점"]

missing = [c for c in needed if c not in df.columns]
#-> 6개의 필드명에 눈에 보이지 않는 화이트 스페이스가 존재할 수 있기 때문에 그 값을 각각 문자의 형태로 가져와서 다시 리스트를 만드는 전처리 진행.
#-> 

if missing : 
    raise ValueError(f"엑셀에서 해당 컬럼을 찾지 못했습니다 : {missing}")

def extract_product_id(url: str) -> int | None : #-> int로 반환하거나 or none
    if not url :
        return None
    m = re.search(r"/products/(\d+)", str(url))  #-> url에서 products뒤에 나오는 1개 이상의 정수를 찾아와라
    return int(m.group(1)) if m else None #-> 가장 처음 나오는 숫자그룹을 정수 형태로 가져와라

def sha1_text(s: str) ->str :
    s = (s or "").strip()
    return hashlib.sha1(s.encode("utf-8")).hexdigest()


    
out = pd.DataFrame()
#데이터 백만개중에 하나라도 값의 형태가 다르면 백만개 다 날라감.따라서 꼼꼼하게 형변환 체크 및 통일시키기
out["brand"] = df["브랜드"].astype(str).str.strip()  #->데이터를 str형태로 바꾸고. 그 해당 str을 찾아와서 strip한다
out["product_name"] = df["상품명"].astype(str).str.strip() 
out["url"] = df["URL"].astype(str).str.strip() 
out["product_id"] = out["url"].apply(extract_product_id)  #-> 해당 함수에 url 컬럼의 각각 값을 적용해라
out["review_count"] = df["리뷰수"]
out["rating"] = df["평점"]
out["review_text"] = (
    df["리뷰"].astype(str)
    .str.replace("\u200b", "", regex=False)  #- whitespace 값 없애기
    .str.replace("\xa0", " ", regex=False)    #- escape sequence 값을 여백 한개로 줄이기
    .str.replace(r"\s+", " ", regex=False)   #- 정규표현식으로 여러개 공백을 1개로 줄이기
    .str.strip()
)
out["review_hash"] = out["review_text"].apply(sha1_text) #-> 중복되는 review값은 같은 코드를 가져서 중복값인지
out["source_file"] = EXCEL_PATH

out = out[out["review_text"].notna() & (out["review_text"] != "")]
out = out.drop_duplicates(subset=["review_hash"]) #-> hash를 기준으로 중복값은 제거해라

# MySQL에 보내기  -> SQL문법으로 써야함

INSERT_SQL = text("""
INSERT INTO musinsa_reviews 
(brand, product_name, product_id, url, review_count, rating, review_text, review_hash, source_file)
VALUES
(:brand, :product_name, :product_id, :url, :review_count, :rating, :review_text, :review_hash, :source_file)
ON DUPLICATE KEY UPDATE
    rating = VALUES(rating),
    review_count = VALUES(review_count),
    product_name = VALUES(product_name),
    brand = VALUES(brand),
    url = VALUES(url),
    source_file = VALUES(source_file);
""")

def chunked(iterable, size: int) :     #-> 데이터가 너무 대량이면 오래걸리니까 덩어리씩 분할해서 보내자
    for i in range(0, len(iterable), size) :
        yield iterable[i:i+size]

#list 형태인 out을 행을 기준으로 하는 딕셔너리 형태로 바꾸기
records = out.to_dict(orient="records")
CHUNK_SIZE = 2000

with ENGINE.begin() as conn :
    for batch in chunked(records, CHUNK_SIZE) :
        conn.execute(INSERT_SQL, batch)

print(f"완료: {len(records)}건 처리완료")
```
