```python
#broadcasting 문법 =  서로다른 행렬의 덴서간 동일한 차원으로 자동 확장 시키고자할때 
```

```python
#서로다른 텐서 중 어느것 하나의 텐서의 차원의 원소수가 1이거나 존재하지 않는 경우 broadcasting 문법을 사용하면 차원을 확장시킬 수 있음
```

```python
data1= torch.FloatTensor([[1],[2],[3]])
data2= torch.FloatTensor([1,2,3])

data3= data1 +data2
print(data1.shape,data2.shape, data3.shape)
print(data3)
```

```python
#서로다른 행렬을 가지고 있는 텐서들 간 연산을 하고자 할때, 다음과 같은 규칙들 중 하나만이라도 부합하면 자동으로 broadcasting 문법 실행 

#1) 서로다른 2개의 tensor의 크기(행과열)가 같은 경우
    #ex) (3,4)(3,4)

#2) 서로다른 2개의 텐서 내 요소값이 1인 경우 
    #ex) (3,4) (1,4)  / 반드시 오른쪽부터 비교. 4==4 조건 성립/ 1->3을 바꾼다  => 결과 (3,4)행렬

data1 = torch.FloatTensor(10,20,4)
data2 = torch.FloatTensor(20,4)

data3 = data1+data2
print(data1.shape,data2.shape,data3.shape)
```
