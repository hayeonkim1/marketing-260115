```python
import numpy as np

data1 = np.array(10)
print(data1, data1.ndim, data1.shape) # scalar 0차원 tensor 
```

```python
import torch

data1 = torch.FloatTensor([1, 2])
print(data1,data1.dim(), data1.shape)  #vector 1차원 tensor
```

```python
torch.zeros(a,b) #모든 원소를 0으로 초기화
torch.ones(a,b) # 모든 원소를 1로 초기화
torch.rand(a, b) #내부 원소들의 값을 0~1사이의 균등분포 방식의 난수를 생성
torch.randn(a, b) #내부 원소들의 값을 평균이 0이고, 표준편차가 1인 정규분포형태의 난수를 생성
torch.full((a, b), c) #내부 원소들의 tensor설정 및 값 초기화
torch.zeros_like() #이미 생성된 tensor와 같은 shape로 0이라는 값을 채우고자 할때
```
