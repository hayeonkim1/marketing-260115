```python
# 데이터 수집 -> 데이터 저장 -> 데이터 구조 -> 데이터 분석
#python > pandas

#pandas : 행과 열을 기반으로 된 테이블 형 데이터를 취급하는 전문 라이브러리 모듈
#데이터를 꼭 테이블 형태로 취급해야하는가? 
  #데이터 취급,분석 주체 = 사람 =익숙한 형태의 구조 = 행과 열 구조 (가로/세로)에 가장 친숙 
#행,열을 다루는데 있어서 다양한 속성, 매서드 기능을 가짐

#데이터 생성 : 오타, 결측치(null), 행/열 위치 바뀜, 필드명 빠져있음 => 상황 수정작업 => 정제(cleaning)
#정제 작업을 포함한 정상적인 데이터 사용을 위한 포괄적 작업 => 전처리    # 전처리는 정제작업을 포함함
#전처리ex. 문자열 타입 -> 날짜타입 / 성별(남여) 표기->0,1  
```

```python
import pandas as pd

#pandas 에서 취급하는 데이터 종류 2가지
#1) 시리즈 series = 1차원 데이터
#2) 데이터 프레임 = dataframe = 2차원데이터

# 각차원 데이터를 표기할때, 유사 리스트의 형태 [] 
#EX) 시리즈는 1차원데이터 -> 1개의 리스트

seriesdata= pd.Series([70,60,70] , index=["국어", "영어", "수학"])  #-> 한줄로 되어있는 1차원데이터에 3개의 값이 들어가있음
#index = 파이썬 프로그래밍 언어, 리스트등의 이터러블한 객체값을 조회 => 인덱스 값 생성
#인덱스를 임의로 지정할 수도 있음

seriesdata
```

```python
seriesdata.index #dtype = int64, object -> 문자열만 있어도 object로 표기, 2개이상의 데이터 타입의 혼합의 경우도 object로 표기
```

```python
#인덱스 변경 가능
seriesdata.index = ["미술", "음악", "체육"]
seriesdata
```

```python
#인덱스 안의 값만 찾아오기
seriesdata.values
```

```python
#특정 인덱스에 해당하는 값을 찾아오기
#1) seiresdata안에 [] 로 찾아오기
seriesdata["미술"]

#2)인덱스 숫자로 찾아오기
print(seriesdata["미술"], seriesdata.iloc[1]) #-> 인덱스의 원래 기존 숫자로 찾아올 수 있음
```

```python
#인덱스 값 대체가능
seriesdata["미술"] = 80
print(seriesdata["미술"])
```

시리즈 데이터 타입 종류

- object: str | 혼용데이터 (문자열 +정수, 정수+논리, 문자열+논리 등등)
- int64 : 정수 (64비트)
- float64 : 부동소수점을 가지고 있는 실수 (64비트)
- bool : True/False
- datetime64 :날짜와 시간 타입

ex) 월 : 1,2,3,4,5,11 월을 문자열로 인식 -> 2월보다 11월을 앞에 정렬함
날짜: 시계열 데이터 -> 시간 경과에 따라 값이 변경되고 있다는 것을 확인할 수 있는 데이터 타입
