```python
df = pd.DataFrame({
    "년도": [2000, 2010, 2020],
    "미국": [2.1, 2.2, 2.3],   
    "한국": [0.4, 0.5, 0.45],
    "중국": [10, 13, 15],
})
```

```python
df
```

```python
#시리즈 값을 인덱스로 변경하기
df.set_index("년도")   
#-> 매서드 함수! 실제 주소를 찾아오는게 아니라 기능을 수행하는 것. 따라서 변경해도 그 주소가 실제로 바뀌는게 아니라 기능에 따라 일시적으로 변경된 값이 출력되는 것 뿐!!
#-> 실제 주소를 변경하려면 df 재할당이 필요함
```
