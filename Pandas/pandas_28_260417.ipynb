```python
#aggregate() => 집계하다 (MySQL)

import pandas as pd

df=pd.DataFrame({
    "미국": [2.1,2.2,2.3],
    "한국": [0.4,0.5,0.45],
    "중국": [10,13,15],
})
df
```

```python
df.aggregate(["sum","min"])
```

```python
df.agg(["sum","min"])
```

```python
df.agg({"미국": "count","한국": "sum"})
```

```python
df.agg("mean", axis=1)  #columns => 열방향 => 행단위로 편균값을 집걔 => 1 /// 0 => 행방향
```

```python
import pandas as pd

df = pd.DataFrame({
    "년도": [2000,2010,2020,2000,2000,2020],
    "미국": [2.1,2.2,2.3,3.1,4.1,5.1],
    "한국": [0.4,0.5,0.45,0.7,0.9,1.2],
    "중국": [10,13,15,12,45,56]    
})
df
```

```python
df.groupby("년도").sum()
```

```python
#년도로 그룹핑했는데 미국의 값만 보고 싶어 =>agg사용 
df.groupby("년도").agg({"미국":"sum"}) 
```

```python
df.groupby("년도").agg({"미국":"sum","한국": "mean"}) 
```

```python
products_category_order = products_category_order.groupby("product_category_name_english").agg({"order_id": "count"}).rename(columns={"order_id":"order_count"})  #order_id에 대한 count값
```

```python
products_category_order
```

```python
products_category_order = products_category_order.sort_values(by = "order_count", ascending = False)
products_category_order.head()
```

```python
import plotly.express as px

fig= px.bar(
    x = products_category_order.index,
    y = products_category_order["order_count"],
    title = "Order Count By Product Category",
    labels = {
        "x": "Product Category",
        "y": "Order Count"
    }
)

fig.update_layout(xaxis_tickangle=-45)

fig.show()
```

```python
import plotly.graph_objects as go

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x = products_category_order.index,
        y = products_category_order["order_count"]
    )
)
fig.update_layout(
    {
        "title": {
            "text": "<b>The Number of Order per Category in Brazilian Olist E-commerce Company</b>",
            "x":0.5,
            "y":0.9,
            "font": {
                "size": 16
            }
        },
        "xaxis" : {
            "title": "From Oct.2016 To Sept. 2018",
            "showticklabels":True,
            "tickfont" : {
                "size": 8 
            }
        },
        "yaxis" :{
            "title":"The number of Order"
        },
        "template" : "plotly_white"
    }
)

fig.show()
```
