```python
#with 절은 무언가를 실행한 후, 해당 행위가 모두 종료되면 알아서 기존에 사용하고 있던 메모리 공간 반납, 종료
with open("00_data/text_data.txt","r",encoding= "utf-8-sig") as file_desc:
    print(file_desc)
```

```python
data_file= open("00_data/text_data.txt","r",encoding= "utf-8-sig")
data_lines = data_file.readlines()
data_lines
```

```python
#데이터를 한줄씩 찾아오기 위한 반복문

for data_line in data_lines :
    print(data_line)
data_file.close()
```

```python
#특정 문서 안에 존재하는 문자열 데이터를 수집할 때, readlines() -> 내부 전체 문자열 모두 찾아옵니다 (단, 리스트 타입)
#-> 따라서 for in 반복문 활용하여 값을 하나씩 출력해와야함.

#readline() -> 가장 최초의 문자열(문장)만 조회, 수집(순수 문자열), =>for in 반복문 활용 x
#동일한 파일을 활용해서 또 다시 readline()-> 이전에 조회 수집한 이후 시점부터 데이터를 찾아온다.
#끝까지 찾아오면 다시 처음으로 돌아가질 않음.
```

```python
#Plain Text 작업 (특정 파일을 불러오거나, 작성하거나, 편집.수정 작업)

#r(ead) : readlines(),  readline(), read()
#w(rite) <- overwrite의 기능 내포 
#a(dd | ppend) 
#        -write: 데이터를 다시 만드는 것
#       -add|append : 데이터에 값을 추가하는 것
```

```python
data_file = open("00_data/text_data_write.txt","w",encoding= "utf-8-sig")
data_file.write("안녕하세요")
data_file.write("Hayeon Kim 입니다.")
data_file.close()
```

```python
#덮어쓰기
data_file = open("00_data/text_data_write.txt","w",encoding= "utf-8-sig")
data_file.write("안녕하세요\n")
data_file.write("Hayeon Kim 입니다.\n")
data_file.close()
```

```python
#문장 추가하기
data_file = open("00_data/text_data_write.txt","a",encoding= "utf-8-sig")
data_file.write("해당 파일은 임시 데이터 입니다.")
data_file.close()
```
