```python
#데이터 분석 : 데이터 찾아와서 수정. 저장
# plain text, csv, xml

#xml = extensible markup language  -> <product></product> 
#html = hyper text markup language ->  <div></div>

from bs4 import BeautifulSoup

data_file = open("00_data/users.xml","r",encoding="utf-8-sig")
# print(data_file)  => 파일의 형태 및 주소를 찾아오는 것 
soup = BeautifulSoup(data_file, "xml") #-> 이게 진짜 그 파일 안의 데이터를 찾아오는 것
users = soup.select("user")
print(users)  
```

```python
#값을 하나씩 찾아오려면 for in 문
from bs4 import BeautifulSoup

data_file = open("00_data/users.xml","r",encoding="utf-8-sig")
# print(data_file)  => 파일의 형태 및 주소를 찾아오는 것 
soup = BeautifulSoup(data_file, "xml") #-> 이게 진짜 그 파일 안의 데이터를 찾아오는 것  #BeautifulSoup을 통해 문자열들을 태그의 형태로 가져올 수 있게함
users = soup.select("user")

for user in users :
    print("나이:", user.select_one("age").text)
    print("이름:", user.select_one("name").text)
```

```python
import requests
from bs4 import BeautifulSoup

keyword = "나이키"
google_related_keyword = f"https://suggestqueries.google.com/complete/search?output=toolbar&q={keyword}"  

#저장된 파일이 아닌 외부의 링크이기 때문에 requsets를 사용해서 그 홈페이지로 get. 
response = requests.get(google_related_keyword)
soup = BeautifulSoup(response.content, "xml")

#태그 찾아오기
datalist = soup.select("suggestion")

for data in datalist :
    print(data["data"])
```
