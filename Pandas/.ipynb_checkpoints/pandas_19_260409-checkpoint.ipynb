```python
#EDA > HEATMAP
import pandas as pd
doc = pd.read_csv("COVID-19-master/csse_covid_19_data/csse_covid_19_daily_reports/04-01-2020.csv", encoding="utf-8-sig")
```

```python
doc
```

```python
doc.corr(numeric_only=True) # correlation (상관관계) coefficient (계수)
```

```python
doc2=doc.corr(numeric_only=True)
```

```python
import plotly.express as px

fig_px = px.imshow(
    doc2,
    text_auto=".2f",
    aspect="auto", #equal
    title="Heatmap (Plotly Express)"
)

fig_px.show()
```

```python
doc2
```

```python
import plotly.graph_objects as go

fig= go.Figure()

fig.add_trace(
    go.Heatmap(
        x=doc2.columns,
        y=doc2.index,
        z=doc2.values,
        colorscale= "Viridis",  #라틴어로 초록이라는 뜻  #Plasma , Cividis, Viridis, Magma, Rainbow
        #colorscale=[(0, "red"), (0.5,"green"), (1,"blue")] -> 아니면 색상코드 삽입해서 수동 설정 가능
        colorbar={
            "title":"Value",
            "thickness":20,
            "tickvals": [-1,0,1],  #구간 정의
            "ticktext": ["음의상관관계","상관관계없음", "양의상관관계"],
            "len": 1
        },
        zmin=-1,
        zmax-1,
        text=doc2.values,
        texttemplate="%{z:.2f}",
        hovertemplate="X:%{x}<br>Y: %{y}<br>Value: %{z:.2f}<extra>Custom Info</extra>" #마우스 오버했을때 x,y값 출력 #<br>: 개행처리 태그
    )
)
fig.update_layout(title="<b>HeatMap (Plotly Graphobjects)</b>")

fig.show()
```
