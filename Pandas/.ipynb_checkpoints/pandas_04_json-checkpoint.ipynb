```python
#plain text, csv, xml, json

#json = Javascript Object Notation
#자바스크립트 객체 = 파이썬 딕셔너리 거의 유사!!
#{name: "david"} -> json /// {"name" : "david"} = 파이썬, 딕셔너리
```

```python
#json형태의 파일을 문자열의 형태로 찾아오는 방법

import json

data = '{"id": "01", "language": "Python", "edition": "third", "author": "David Park"}' #->json은 반드시 문자열이어야 하므로 ''

jsondata = json.loads(data)
print(jsondata["id"],jsondata["language"],jsondata["edition"])
print(type(jsondata))
```

```python
#json 형태의 파일을 실제 본 형태를 유지한 채로 찾아오는 방법
import json

data = {"id": "01", "language": "Python", "edition": "third", "author": "David Park"}
jsondata = json.dumps(data) 
print(jsondata, type(jsondata))
```

```python
import json

data = {"id": "01", "language": "Python", "edition": "third", "author": "David Park"}

with open("00_data/test.json", "w", encoding="utf-8-sig") as json_file :
    json_string = json.dump(data, json_file, indent=2)   #indent: 들여쓰기 (=2 : 2칸만큼)
    
```
