```python
#csv파일을 리스트 형태로 값을 삽입하는 방법
mport csv

with open("00_data/tmp_csv.csv","w",encoding="utf-8-sig") as writer_csv :
    writer = csv.writer(writer_csv, delimiter=",")
    writer.writerow(["love"] * 3 + ["banana"])  #-> 한 행안에 love를 3번 넣어라 (1열,2열,3열)
    writer.writerow(["apple",2])
    writer.writerow(["dog","cat","mouse"])
```

```python
#행.열 기반의 테이블 안에 어떤값을 넣을 때, 1개의 행을 구분짓는 자료형 => 리스트([])
```

```python
#csv 파일을 딕셔너리 형태로 값을 삽입하는 방법
import csv

with open("00_data/tmp_csv.csv","w",encoding="utf-8-sig") as writer_csv :
    field_name_list = ["First_Name", "Last_Name"] #-> 첫번째 행의 값을 미리 지정해서 넣기
    writer = csv.DictWriter(writer_csv, fieldnames= field_name_list)
    writer.writeheader() #-> 최상단의 행을 헤더로 설정하게 함
    writer.writerow({"First_Name": "David", "Last_Name":"Park"}) #-> 그 각각의 열에 값 넣기
    writer.writerow({"First_Name": "Dave", "Last_Name":"Lee"}) #-> 그 각각의 열에 값 넣기
    writer.writerow({"First_Name": "Robin", "Last_Name":"Hood"}) #-> 그 각각의 열에 값 넣기
    
    
```
