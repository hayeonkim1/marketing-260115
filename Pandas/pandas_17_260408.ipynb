```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go

df = pd.DataFrame(np.random.rand(10,2), columns = ["A","B"])

fig = go.Figure()  #인스턴스 객체 생성

fig.add_trace (
    go.Bar(
        x= df.index, y= df["A"], name="A", text=df["A"], textposition="auto"
    )
)
fig.add_trace (
    go.Bar(
        x= df.index, y= df["B"], name="B", text=df["B"], textposition="auto"
    )
)

fig.show()
```

```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go

df= pd.DataFrame({
    "A": np.random.randint(10,50, size=10),  # 디폴트 값: 0-9 사이의 정수를 난수형태로 생성 #생성범위 설정 가능 #size: 갯수
    "B": np.random.randint(10,50, size=10),
})

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=df.index, y=df["A"], name="A", text=df["A"], textposition="auto", texttemplate="%{y:.32f}"  # 소숫점 2번째 자리까지만 표기
    )
)
fig.add_trace(
    go.Bar(
        x=df.index, y=df["B"], name="B", text=df["B"], textposition="auto",texttemplate="%{y:.2f}"
    )
)

fig .update_layout(
    {
        "title":{
            "text":"Graph with <b>go.Bar</b>",  #<b></b> : 볼드처리
            "x": 0.5, 
            "y": 0.9,
            "font": {
                "size":20
            }
        },
        "showlegend" : True,
        "xaxis": {
            "title": "index number",
            "showticklabels" :True,
            "dtick": 1
        },
        "yaxis" :{
            "title": "randint"
        },
        "autosize"  :True #반응형 그래프 : 화면에 따라 너비나 높이를 자동으로 줄이거나 늘림
    }    
)

fig.show()
```
