```python


from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys #enter 키 누루는 코드
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager
import time
import csv
import re 

URL = "https://www.ssg.com/"

service = Service(ChromeDriverManager().install())
options = Options()

options.add_argument("--disable-dev-shm-usage")
options.add_argument("--window-size=1920,1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent= Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36 ")
options.add_argument("--lang=ko_KR")
options.add_argument("--no-sandbox")

driver = webdriver.Chrome(service=service, options=options)
driver.get(URL)
time.sleep(3)

ssgmall_tab = driver.find_element(By.XPATH, "//a[contains(@class, 'gnb_mall_link') and contains(text(),'신세계몰')]")
ssgmall_tab.click()
time.sleep(3)

ssgspecial_tab = driver.find_element(By.XPATH, "//a[contains(@class, 'menu_lnk clickable') and contains(text(),'쓱-특가')]")
ssgspecial_tab.click()
time.sleep(3)

food_button = driver.find_element(By.XPATH, "//button[@data-index='8']")
food_button.click()
time.sleep(3)

results = list()


#이스케이프 시퀀스 제거를 위한 함수 정의 -> 모든 변수에 일일히 하기 싫으니까 함수로 정의해서 한번에

def clean_txt(s) : 
    if not s : 
        return ""
    s = re.sub(r"\s+", " ", s)
    return s.strip()


while len(results) < 50 :
    cards = driver.find_elements(By.CSS_SELECTOR, "div.template-grid-item.css-8kawlz")

    for index, card in enumerate(cards) : 
        try :
            a = card.find_element(By.CSS_SELECTOR, "a[href]")
            url = a.get_attribute("href")
        except :
            continue
            
        try :
            name = card.find_element(By.CSS_SELECTOR, "p.chakra-text.css-19bfb2a").text.strip()
            name = clean_txt(name)
        except :
            name = ""

        saleAndprice = card.find_element(By.CSS_SELECTOR, "div.chakra-stack.css-ffjhre")

        try:
            discount = saleAndprice.find_element(By.CSS_SELECTOR,"em.css-aywnvu").text.strip()
            discount = clean_txt(discount).replace("할인율", "").strip()
        except :
            discount = ""
            
        try:
            price = saleAndprice.find_element(By.CSS_SELECTOR,"em.css-1oiygnj").text.strip()
            price = clean_txt(price).replace("판매가격","").strip()
        except :
            price = ""

            
        results.append({"No": index+1, "상품명": name, "판매금액": price, "할인율": discount, "URL": url})
        


        


driver.quit()

results
```

위 처럼 하면 1-38번까지 크롱링 후 1번부터 다시 돌아서 50개를 채움
=> set을 사용해서 중복값을 제외
=> 가장 일치하거나 값이 없을 확률이 적은 url값을 통해 중복값 감지.

```python


from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys #enter 키 누루는 코드
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager
import pandas as pd
import time
import csv
import re 

URL = "https://www.ssg.com/"

service = Service(ChromeDriverManager().install())
options = Options()

options.add_argument("--disable-dev-shm-usage")
options.add_argument("--window-size=1920,1080")
options.add_argument("--start-maximized")
options.add_argument("--user-agent= Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.36 ")
options.add_argument("--lang=ko_KR")
options.add_argument("--no-sandbox")

driver = webdriver.Chrome(service=service, options=options)
driver.get(URL)
time.sleep(3)

ssgmall_tab = driver.find_element(By.XPATH, "//a[contains(@class, 'gnb_mall_link') and contains(text(),'신세계몰')]")
ssgmall_tab.click()
time.sleep(3)

ssgspecial_tab = driver.find_element(By.XPATH, "//a[contains(@class, 'menu_lnk clickable') and contains(text(),'쓱-특가')]")
ssgspecial_tab.click()
time.sleep(3)

food_button = driver.find_element(By.XPATH, "//button[@data-index='8']")
food_button.click()
time.sleep(3)

results = list()
seen_urls = set()

#이스케이프 시퀀스 제거를 위한 함수 정의 -> 모든 변수에 일일히 하기 싫으니까 함수로 정의해서 한번에

def clean_txt(s) : 
    if not s : 
        return ""
    s = re.sub(r"\s+", " ", s)
    return s.strip()


while len(results) < 50 :
    cards = driver.find_elements(By.CSS_SELECTOR, "div.template-grid-item.css-8kawlz")
    added_this_round = 0
    
    
    for index, card in enumerate(cards) : 
        try :
            a = card.find_element(By.CSS_SELECTOR, "a[href]")
            url = a.get_attribute("href")
        except :
            continue
            
        try :
            name = card.find_element(By.CSS_SELECTOR, "p.chakra-text.css-19bfb2a").text.strip()
            name = clean_txt(name)
        except :
            name = ""

        saleAndprice = card.find_element(By.CSS_SELECTOR, "div.chakra-stack.css-ffjhre")

        try:
            discount = saleAndprice.find_element(By.CSS_SELECTOR,"em.css-aywnvu").text.strip()
            discount = clean_txt(discount).replace("할인율", "").strip()
        except :
            discount = ""
            
        try:
            price = saleAndprice.find_element(By.CSS_SELECTOR,"em.css-1oiygnj").text.strip()
            price = clean_txt(price).replace("판매가격","").strip()
        except :
            price = ""

        #set를 거친 값만 리스트에 추가되게함
        if url in seen_urls :
            continue
        seen_urls.add(url)

            
        results.append({"No": index+1, "상품명": name, "판매금액": price, "할인율": discount, "URL": url})
        added_this_round += 1

    if added_this_round == 0 :
        print("새 상품이 더이상 로드되지 않습니다.")
        break
   

df = pd.DataFrame(results).head(50)
driver.quit()

df
```
