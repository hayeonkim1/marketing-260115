```python
from selenium import webdriver #--> webdriver가 실질적 크롤러 역할을 함
from selenium.webdriver.chrome.service import Service #웹 브라우저 드라이브 이식하는 역할
from selenium.webdriver.chrome.options import Options #웹의 옵션들을 지정
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
from urllib import parse 
import time
import pandas as pd #--> as 뒤의 별칭으로 부르겠다

service = Service(ChromeDriverManager().install()) #--> 대문자이고 복잡한 클래스 대신 변수로 지정해서 사용
options = Options()
# options.add_argument("--headless")
# options.add_argument("--disable-gpu")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--window-size=190*1080")
options.add_argument("--start-maximized") 
options.add_argument("--user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/143.0.0.0 Safari/537.36")
options.add_argument("--lang=ko_KR")

driver = webdriver.Chrome(service=service, options=options)


query = "마케팅" 
encoded_query = parse.quote(query)



max_pages = 1

postings = []  #-> 전체 공고제목 &지원링크를 한 list 안에 넣기 /// -> 리스트 값은 나중에 append()로 뒤에서부터 넣을 거임.
for page in range(1, max_pages + 1) : 
    url = f"https://www.jobkorea.co.kr/Search?stext={encoded_query}&Page_No={page}"
    driver.get(url)
    time.sleep(2) 
    
    # job_posts = driver.find_elements(By.CSS_SELECTOR, "div.styles_mb_space2__dk46ts4t") 
    job_posts = driver.find_elements(By.CSS_SELECTOR, "div.Box_bgColor_white__1wwr54u0.Box_borderColor_default__1wwr54u5.Box_borderSize_1__1wwr54ud.styles_p_space0__dk46ts61.styles_radius_radius16__dk46ts9d.Shadow_root_list__bm2zcc6.dlua7o0" )
    for post in job_posts :
        try : # 정상적으로 실행될 경우 아래를 진행
            
                
            title = post.find_element(By.CSS_SELECTOR, "div.styles_mb_space2__dk46ts4t").text.strip() 
            company = post.find_element(By.CSS_SELECTOR, "span.Typography_variant_size16__344nw26.Typography_weight_regular__344nw2e.Typography_color_gray700__344nw2o.Typography_truncate__344nw2y").text.strip() 
            location = post.find_element(By.CSS_SELECTOR ,"div.Flex_display_flex__i0l0hl2.Flex_gap_space4__i0l0hly.Flex_align_center__i0l0hl8.styles_flex-shrink_0__dk46tsa9.styles_px_space10__dk46tsm._10fvqwx0 > span.Typography_variant_size14__344nw27.Typography_weight_regular__344nw2e.Typography_color_gray900__344nw2m.Typography_truncate__344nw2y").text.strip()
            keyword = post.find_element(By.CSS_SELECTOR, "div.Flex_display_flex__i0l0hl2.Flex_gap_space4__i0l0hly.Flex_align_center__i0l0hl8.styles_flex-shrink_1__dk46tsaa.styles_px_space10__dk46tsm._10fvqwx0 > span.Typography_variant_size14__344nw27.Typography_weight_regular__344nw2e.Typography_color_gray900__344nw2m.Typography_truncate__344nw2y ").text.strip()
                                         
            # link = title_elem.get_attribute("href") #--> 공고 url을 가져옴
            postings.append({
                "모집명": title,
                "모집 기관": company,
                "지역": location,
                "키워드": keyword
            })
            
        except : #실행 안될 경우 아래 continue를 실행
            print("오류발생")

print(postings)
    
driver.quit()
```

```python
[오늘의 미션] 3/20

잡코리아 사이트 접속 후 상단 검색바 "마케팅" 입력 > 출력되는 결과페이지 (1페이지에 한함)
데이터 크롤링 (광고는 제외, 모집명, 모집기관, 지역, 키워드)까지 수집
MongoDB NoSQL 데이터 저장
```

```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017")

#값 생성
db = client["jobkorea"]
contents = db.contents
#컬렉션에 값넣기
data = postings
result = contents.insert_many(data)

print(result)
```
