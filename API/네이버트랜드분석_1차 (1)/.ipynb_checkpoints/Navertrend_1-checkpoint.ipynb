```python
import urllib.request
import json
import pandas as pd
import plotly.express as px

client_id = "iQYuA4jIAZWE78ZCH4p9" 
client_secret = "7MP3lY3dzz"
url = "https://openapi.naver.com/v1/datalab/search"

# 1. 분석 설정
target_kw = "가성비영양제" # 가장 핵심인 키워드 하나를 기준으로 연령 비교
age_dict = {'20대': ['3', '4'], '30대': ['5', '6'], '40대': ['7', '8']}
all_data = []

# 2. 연령별 루프 실행 (연령별로 API를 따로 호출하여 수집)
for age_name, age_codes in age_dict.items():
    body = {
        "startDate": "2024-01-01",
        "endDate": "2026-05-14",
        "timeUnit": "month",
        "keywordGroups": [{"groupName": target_kw, "keywords": [target_kw]}],
        "ages": age_codes
    }
    
    request = urllib.request.Request(url)
    request.add_header("X-Naver-Client-Id", client_id)
    request.add_header("X-Naver-Client-Secret", client_secret)
    request.add_header("Content-Type", "application/json")
    
    response = urllib.request.urlopen(request, data=json.dumps(body).encode("utf-8"))
    res_data = json.loads(response.read())
    
    # 데이터 정리
    for item in res_data['results'][0]['data']:
        all_data.append({
            'period': item['period'],
            'ratio': item['ratio'],
            'age_group': age_name
        })

# 3. 데이터프레임 통합
df_age = pd.DataFrame(all_data)

# -----------------------------------------------------------
# 🎯 [CSV 저장 로직] 독립된 CSV 파일로 저장하기
# -----------------------------------------------------------
file_name = "가성비영양제_연령별_검색트랜드.csv"
df_age.to_csv(file_name, index=False, encoding="utf-8-sig")
print(f"💾 {file_name} 파일이 성공적으로 저장되었습니다!")
# -----------------------------------------------------------

# 4. 시각화 (연령별 비교)
fig = px.line(df_age, x='period', y='ratio', color='age_group',
              title=f"'{target_kw}' 연령별 검색 트렌드 비교",
              labels={'ratio': '상대적 검색량', 'age_group': '연령대'},
              markers=True, template='plotly_white')

fig.show()
```

```python
import os
import sys
import urllib.request
import json
import pandas as pd
import plotly.express as px

# 1. 네이버 개발자 센터에서 발급받은 키 입력
client_id = "iQYuA4jIAZWE78ZCH4p9" 
client_secret = "7MP3lY3dzz"
keyword = "가성비영양제"

# 2. 네이버 블로그 검색 API 호출
encText = urllib.parse.quote(keyword)
url = f"https://openapi.naver.com/v1/search/blog?query={encText}&display=100"

request = urllib.request.Request(url)
request.add_header("X-Naver-Client-Id", client_id)
request.add_header("X-Naver-Client-Secret", client_secret)

try:
    response = urllib.request.urlopen(request)
    if response.getcode() == 200:
        data = json.loads(response.read().decode('utf-8'))
        total_items = data['total']
        print(f"네이버 검색 결과 총 건수: {total_items:,}건")
        
        # 3. 데이터프레임 변환 및 날짜 보정
        df = pd.DataFrame(data['items'])
        df['postdate'] = df['postdate'].astype(str)
        df['postdate'] = pd.to_datetime(df['postdate'], format='%Y%m%d', errors='coerce')
        
        # 2026년 1월 1일 이후 데이터 필터링
        df_filtered = df[df['postdate'] >= '2026-01-01'].copy()
        print(f"최근 수집된 100개 중 2026년 이후 작성된 글: {len(df_filtered)}건")
        
        # 4. 시각화 데이터 가공 및 그래프 출력
        if not df_filtered.empty:
            df_count = df_filtered['bloggername'].value_counts().reset_index()
            df_count.columns = ['블로그 출처', '게시글 수']
            
            fig = px.bar(df_count.head(15), x='블로그 출처', y='게시글 수', 
                         title=f"2026년 이후 '{keyword}' 관련 주요 출처", 
                         text='게시글 수', template='plotly_white', 
                         color_discrete_sequence=['#652D90'])
            
            # [수정 완료] 'outer'에서 'outside'로 변경하여 에러를 해결했습니다.
            fig.update_traces(textposition='outside')
            
            fig.show()
        else:
            print("2026년 이후에 발행된 글이 수집 범위 내에 없습니다.")
    else:
        print("API 요청 실패 코드:", response.getcode())
except Exception as e:
    print(f"에러 발생: {e}")
```

```python
import os
import sys
import urllib.request
import json
import pandas as pd
import plotly.express as px

# 1. 인증 정보 설정
client_id = "iQYuA4jIAZWE78ZCH4p9" 
client_secret = "7MP3lY3dzz"
url = "https://openapi.naver.com/v1/datalab/search"

# 2. 분석 키워드 설정
target_keywords = ["가성비 영양제"]

# 3. 키워드별 개별 그룹 자동 생성
keyword_groups = []
for kw in target_keywords:
    keyword_groups.append({
        "groupName": kw, 
        "keywords": [kw] 
    })

# 4. API 요청 본문 작성 (2023년부터 2026년까지)
body = {
    "startDate": "2023-01-01",
    "endDate": "2026-05-14",
    "timeUnit": "month",
    "keywordGroups": keyword_groups
}
body_json = json.dumps(body, ensure_ascii=False)

# 5. API 호출 실행
request = urllib.request.Request(url)
request.add_header("X-Naver-Client-Id", client_id)
request.add_header("X-Naver-Client-Secret", client_secret)
request.add_header("Content-Type", "application/json")

try:
    response = urllib.request.urlopen(request, data=body_json.encode("utf-8"))
    rescode = response.getcode()
    
    if rescode == 200:
        response_body = response.read()
        data = json.loads(response_body)
        
        # 6. 데이터프레임 변환 (final_df3 생성)
        df_list = []
        for res in data['results']:
            temp_df = pd.DataFrame(res['data'])
            temp_df['group'] = res['title']
            df_list.append(temp_df)
        
        cheap_real_search_count = pd.concat(df_list, ignore_index=True)
        print("📊 네이버 데이터랩 수집 완료 ( cheap_real_search_count 생성 성공)")
        
        # -----------------------------------------------------------
        # 7. final_df3를 활용한 연도별 진짜 트렌드 환산 알고리즘
        # -----------------------------------------------------------
        df_trend = cheap_real_search_count.copy()
        df_trend['year'] = df_trend['period'].str[:4] # '2024-01' -> '2024'
        
        # 블로그 API로 확인하셨던 실측 누적 기준 수치
        recent_actual_volume = 242543  
        
        # [안전장치] NaN(빈값)을 완전히 제외하고, 실제 숫자가 들어있는 행들만 골라냅니다.
        df_valid = df_trend[(df_trend['group'] == "가성비 영양제") & (df_trend['ratio'].notna())]
        
        # 실제 데이터가 존재하는 가장 최신 시점의 행을 안전하게 확보합니다.
        if not df_valid.empty:
            last_valid_row = df_valid.sort_values('period').iloc[-1]
            recent_ratio = last_valid_row['ratio']
            print(f"🎯 안전 기준점 감지 완료 -> 기준 달: {last_valid_row['period']}, 기준 비율(ratio): {recent_ratio}")
        else:
            recent_ratio = 50.0
            
        # 비율(ratio) 1당 실제 문서 건수 계산
        ratio_to_volume_constant = recent_actual_volume / recent_ratio
        
        # 월별 상대적 비율에 상수를 곱해 '진짜 발행 건수' 역산
        df_trend['actual_volume'] = (df_trend['ratio'] * ratio_to_volume_constant).round(0)
        df_trend['actual_volume'] = df_trend['actual_volume'].fillna(0)
        
        # 연도별로 그룹화하여 '연간 총 발행 건수' 최종 합산
        df_yearly_total = df_trend.groupby('year')['actual_volume'].sum().reset_index()
        df_yearly_total.columns = ['연도', '연간 총 발행건수(추정)']
        
        df_yearly_total['연도'] = df_yearly_total['연도'].astype(str)
        df_yearly_total = df_yearly_total.sort_values('연도')
        
        print("\n--- 💸 연도별 진짜 발행 건수 최종 합산 결과 ---")
        print(df_yearly_total)
        
        # -----------------------------------------------------------
        # 🎯 [요청 기능 추가] 모든 연도수의 누적 합계값 데이터프레임 생성
        # -----------------------------------------------------------
        total_sum_val = df_yearly_total['연간 총 발행건수(추정)'].sum()
        df_total_sum = pd.DataFrame([{
            '구분': '전체 기간 총 누적 합계',
            '총 발행건수(추정)': total_sum_val
        }])
        
        print("\n--- 🏆 모든 연도 누적 합계 결과 (df_total_sum) ---")
        print(df_total_sum)
        print(f"👉 전체 기간 동안 총 {total_sum_val:,.0f}건의 문서가 발행된 것으로 추정됩니다.\n")
        
     
        
        # 🎯 [CSV 저장 로직] 연도별 총 발행건수 데이터만 CSV 파일로 출력
        # -----------------------------------------------------------
        file_yearly = "가성비영양제_실검색건수추이.csv"
        
        # 한글 안깨지게 utf-8-sig 인코딩 부여 및 무의미한 인덱스 열 제거
        df_yearly_total.to_csv(file_yearly, index=False, encoding="utf-8-sig")
        print(f"💾 {file_yearly} 파일 저장 완료!\n")
        # -----------------------------------------------------------
        
        # -----------------------------------------------------------
        # 8. 연도별 총 수치 시각화
        # -----------------------------------------------------------
        fig_bar = px.bar(df_yearly_total, 
                         x='연도', 
                         y='연간 총 발행건수(추정)',
                         title="연도별 '가성비 영양제' 관련 총 발행 건수 추이",
                         text='연간 총 발행건수(추정)', 
                         template='plotly_white')
        
        fig_bar.update_traces(marker_color='#652D90', # CU 퍼플
                              texttemplate='%{text:,}건', 
                              textposition='outside')
        
        fig_bar.update_layout(xaxis_type='category',
                              yaxis_title="발행 건수 (건)",
                              xaxis_title="연도")
        
        fig_bar.show()
    else:
        print(f"Error: API 요청 실패 (Code: {rescode})")

except Exception as e:
    print(f"Error 발생: {e}")
```
