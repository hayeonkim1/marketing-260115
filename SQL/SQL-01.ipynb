# 데이터 저장을 위해 알고있어야하는 중요한 개념

1. 데이터 (Data)
   - 관찰, 측저으 조사 등을 통해 수집된 사실, 값, 기록등의 원초적인 정보
   - 데이터 그 자체로는 의미부여가 많이 부족함
   - 데이터가 가치를 가지려면, 쌓이고 축적된 구조화된 데이터 비로소 의미있는 가치 부여

2. 데이터베이스 (Database = DB)
   - 다양한 데이터를 '구조적'이고, '체계적'으로 저장 및 관리할 수 있는 저장소
2가지 종류
(1) 관계형 데이터베이스 : RDBMS -> 구조화된 다양한 데이터들이 서로 관계를 맺은 형태
> 오랜 역사를 가지고 있음
> 대기업, 글로벌 기업이 주로 사용
> 안정적, 검증된 데이터 관리시스템
> 약점: 데이터 간 관계를 위한 기본적으로 사용해야하는 문법체계존재 => 문법 의존도 높음
> 데이터 우선?  vs 데이터를 다루는 언어 문법 체계가 우선?  대립발생
> Oracle, MySQL
> SQLD

(2) 비관계형 데이터베이스 : CRUD -> 구조화된 데이터들의 관계에 치중하지 않고 데이터의 입출력 및 수정,편집등을 원활하게 하기 위한 시스템
> CRUD:
> Create / Read/ Update / Delete
> 스타트업| 데이터 입출력이 빈번한 글로벌기업
> MongoDB
>NoSQL => Not only SQL
> 유연한 문법체계 = 데이터에 포커스

3. 스키마 (Schema / Scheme)
    : 데이터를 체계적으로 구조화 한 것이 데이터 베이스 -> DB생성을 위해 어떤 규칙과 패턴을 가지고 각 데이터들을 체계적으로 구조화 했는가
   - 데이터를 데이터베이스화 하기 위한 목적으로 어떤 약속된 구조(틀)을 정의해놓은 것.
   - 전체 몇개의 컬럼을 기준으로 구조화할 것인가?
   - 각 컬럼안에 입력될 데이터 값의 타입은?
   - 데이터의 타입을 세부화해서 메모리를 얼만큼 가용하게 할 것인가?

     
4. 테이블(Table)
   - 다량의 데이터를 모두 독립적, 개별적으로 사용 => 비효율적
   - 비슷한 유형의 데이터 (ex. 1/10은 소비자 데이터, 2/10은 생산자데이터
   - 데이터 ㅜ조화를 위한 데이터 베이스 관리 시스템 적용 -> 다양한 데이터를 유사한 속성에 따라 행,열 기반의 테이블로 그룹화, 데이터베이스 안에 그룹별 저장 가능
   - 실질적 데이터 저장 단위
   - 하나의 DB에는 복수의 테이블 존재 가능

5. SQL(Structured Query Language) = 관계형 데이터 베이스를 다루기 위한 문법 체계
   - DDL(Data Definition Language) : 데이터 생성, 삭제, 교체
   - DML(Data Manipulation Language) : 데이터 조회, 삽입, 업데이트
   - DCL(Data Control Langugae) : 데이터 권한, 박탈, 등급

6. MySQL 학습

    -명령프롭포트 기반 실행

   -GUI : 워크벤치

# 테이블 생성시 꼭 이해해야하는 개념
- 데이터베이스 생성 > 테이블 생성 > 데이터 저장
- 좋은 테이블 생성 = 최대한 데이터를 효율적으로 저장

- 최대한 중복되는 값을 피한다.
      -> 테이블 생성시 각각의 로우| 레코드를 식별할 수 있는 고유 키를 생성함
      => Primary Key =기본키
  - 다른 외부 데이터와의 호환성 = 쉬운 관계성
        => Foreign Key = 외부키 

```python
1. 데이터 베이스 생성

2. 현재 MySQL안에 생성된 데이터 베이스 조회

3. 데이터베이스 선택

4. 선택된 데이터 베이스안에 테이블 생성
ㅇㄹㄴㄹf
5. 데이터베이스 안에 생성된 테이블 조회
```

```python
-- 주석1: 
# 주석 2: 단문주석 -> MySQL 전용주석
/* 주석 3 : 복문주석 */
-- 반드시 SQL문법을 사용할때에는 종료구문에 세미코론 ;

/*
MySQL 접속시 이것부터 시작해라!

1. 데이터베이스 생성
CREATE DATABASE <dbname>;
> 데이터 베이스의 이름은 항상 직관적, 명시적 -> 누가봐도 한눈에 알아볼 수 있도록

CREATE DATABASE IF NOT EXISTS <dbname>;
-> 실수로 덮어쓰기 해버리면 그전 데이터 복구 불가.
-> 이렇게 하는게 안전함

CREATE DATABASE IF NOT EXISTS <dbname> 
CHARACTER SET utf8mb4 COLLATE utf8mb4_general_ci; 
이모지 등의 다양한 형태의 데이터가 깨지지않고 들어오게끔
-> 세미콜론만 없으면 줄바꿈해도 하나의 이어지는 코드로 인식
->UTF :  Unicode Transformation Format
각 나라의 언어는 국제적으로 약속된 코드형식으로 이루어져있음



2. 현재 MySQL 안에 생성된 데이터베이스 조회
SHOW DATABASES;
-> 왼쪽 schemas 에 볼드 처리가 되야 조회된 것. 

3. 데이터베이스 선택
USE <dbname>;
4. 선택된 데이터 베이스안에 테이블 생성

5. 데이터베이스 안에 생성된 테이블 조회 

6. 생성된 테이블 안에 데이터 저장.삽입

7. 데이터베이스 삭제
DROP DATABASE <dbname>;
> DROP DATABASE IF EXISTS <dbname>;


ctrl + enter : 단문실행
ctrl + shift + enter : 복문실행

*/

CREATE DATABASE digitalmkt;
CREATE DATABASE IF NOT EXISTS digitalmkt;

CREATE DATABASE IF NOT EXISTS digitalmkt
CHARACTER SET utf8mb4
COLLATE utf8mb4_general_ci;

SHOW DATABASES;
USE digitalmkt;
DROP DATABASE digitalmkt;
DROP DATABASE IF EXISTS digitalmkt;
```
