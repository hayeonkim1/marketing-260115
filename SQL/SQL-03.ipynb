```python
import pymysql 
#mysql에서 db 만들고 오기

#기본 세팅
db= pymysql.connect(
    host="localhost", #127.0.0.1를 써도 됨
    port=3306,
    user="root",
    passwd="Hayeon123!",
    db="ecommerce",
    charset="utf8mb4"
)
```

```python
db
```

```python
#정상적으로python &mysql 연결 후 
#cursor() 매서드 호출
# => Cursor 객체 -> mysql DB안에 값을 전송해주는 역할

ecommerce = db.cursor()  #-> cursor() 객체를 사용해서 ecommerce라는 db안에 값을 전송
```

```python
ecommerce
```

```python
파이썬에서 sql문법으로 테이블 생성하기

# sql 문법은 """ """ 큰따옴표로 개행 처리를 해야 가능함. + ; 종료
```

```python
sql = """ 
    CREATE TABLE product (
        product_code VARCHAR(20) NOT NULL PRIMARY KEY,
        title VARCHAR(200) NOT NULL,
        ori_price INT,
        discount_price INT,
        discount_percent INT,
        delivery VARCHAR(2)
    );
"""
```

```python
sql
```

```python
#mysql에서 실행할 수 있게 하기
ecommerce.execute(sql)  #execute : 실행
```

```python
#실행한 결과값 기록->  commit을 해줘야 기록이 됨. 
db.commit()  

#실행하고 sql가보면 table이 생성되어 있음 But 아직 값은 없음
```

```python
db.close()
```

db 연결한 이후 close() => 오픈됐던 db 종료
만약 다시 동일한 db의 테이블에 접속하려면 다시 db connect를 해야함!!

```python
cursor = db.cursor()
SQL = "SELECT * FROM product;"
cursor.execute(SQL)


# row = cursor.fetchone() #값을 찾아오는것 =  fetch
row = cursor.fetchmany(2) #-> tuple 형태로 값을 찾아온다 => 반복문 사용가능
print(row)

row = cursor.fetchone() # 앞에서 찾아온값 그 다음 값을 찾아오라는 것
print(row)

db.close()
```
